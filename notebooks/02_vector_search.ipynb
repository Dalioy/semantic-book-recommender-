# 02 · Vector Search

Builds the Chroma vector store over book descriptions (embedded locally via Ollama's `nomic-embed-text`), then runs a sample semantic search.

**Prerequisite:** `ollama serve` running, with the embedding model pulled: `ollama pull nomic-embed-text`.

## Setup

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

from src.config import CLEANED_BOOKS_CSV
from src.vector_store.build_index import (
    load_tagged_documents, build_chroma_index, load_chroma_index,
)
from src.recommender.retrieval import retrieve_semantic_recommendations
```

## 1. Load the cleaned books + tagged descriptions

```python
books = pd.read_csv(CLEANED_BOOKS_CSV)
documents = load_tagged_documents()
documents[0]
```

## 2. Build the vector index

This embeds every book description in batches and persists the result to `chroma_db/` at the project root.

```python
db_books = build_chroma_index(documents)
```

## 3. Reload the index (sanity check)

Confirms the persisted collection loads back correctly — useful if you're resuming work without re-embedding.

```python
db_books = load_chroma_index()
db_books._collection.count()
```

## 4. Try a semantic search

```python
retrieve_semantic_recommendations(db_books, books, query="horror")
```
